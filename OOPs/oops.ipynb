```python
# inheritance
Inheritance is a mechanism in object-oriented programming that allows a new class (called a subclass or derived class)
to inherit attributes and methods from an existing class (called a superclass or base class). 
This promotes code reusability and establishes a hierarchical relationship between classes.
```

```python
class person:
    def __init__(self,name,age,gender):
        self.name=name  #attribute
        self.age=age
        self.gender=gender
    def introduce(self):  # Method
        print(f"Hi I am {self.name} and my age is {self.age} and gender is {self.gender}")
        # print(f"{self.role}")
class child(person):  # inheritance
    def __init__(self,name,age,gender,role,salary):
        self.name=name
        self.age=age
        self.gender=gender
        self.role=role
        self.salary=salary
        super().__init__(name,age,gender) #parent class constructor
        #
    def play(self):
        print(f"My Name is {self.name} and I am {self.age} years old. I love to playyyy")
        print(f"My Role is {self.role} and my salary is {self.salary}")
    def introduce(self):  # Method
        print(f"Hi I am {self.name} and my age is {self.age} and gender is {self.gender}")
        # print(f"{self.role}")
s1=child("Vishnu",21,"m","Developer",50000)
s1.introduce()
s2=person("Madhav",50,"m")
s2.introduce()
s1.play()

# s1=person("Madhav",22)
# s1.play()
```

```python
class phone:
    def call(self,number):
        print(f"Calling {number}...")
class Camera:
    def take_photo(self):
        print("Photo Taken")
class Smartphone(phone,Camera):  # multiple inheritance
    def browse_internet(self,url):
        print(f"Browsing {url}...")
s1=Smartphone()
s1.call(1234567890)
s1.take_photo()
s1.browse_internet("www.example.com")
```

```python
class user:
    def __init__(self,username,email):
        self.username=username
        self.email=email
        self.__login=False
    def login(self,password):
        if password=="admin123":
            self.__login=True
            print(f"{self.username} logged in successfully")
        else:
            print("Invalid Credentials")
    def logout(self):
        self.__login=False
        print(f"{self.username} logged out successfully")
    def is_logged_in(self):
        return self.__login
class seller(user):  # inheritance
    def __init__(self,username,email,shop_name):
        super().__init__(username,email)
        self.shop_name=shop_name
        self.products=[]
    def add_product(self,product):
        if self.is_logged_in():
            self.products.append(product)
            print(f"Product {product} added to shop {self.shop_name}")
        else:
            print("Login Required to add products")
    def sell(self,product):
        if self.is_logged_in():
            if product in self.products:
                self.products.remove(product)
                print(f"Product {product} sold from shop {self.shop_name}")
            else:
                print(f"Product {product} not found in shop {self.shop_name}")
        else:
            print("Login Required to sell products")
class premium_seller(seller):
    def __init__(self,username,email,shop_name,ads_budget):
        super().__init__(username,email,shop_name)
        self.ads_budget=ads_budget
    def run_ads(self):
        if self.ads_budget>0:
            print(f"Running ads for shop {self.shop_name} with budget {self.ads_budget}")
        else:
            print("No budget to run ads")
    def sell(self,product):
        super().sell(product)# method overriding(Polymorphism)
        # print("Priority Listing Enabled for premium sellers")
        
ps=premium_seller(
    username="vishnu",
    email="vishnu.klhs@gmail.com",
    shop_name="Vishnu's Store",
    ads_budget=0
)
ps.login("admin123")
ps.add_product("Laptop")
ps.sell("Laptop")
# ps.sell("Laptop")
ps.run_ads()
```
