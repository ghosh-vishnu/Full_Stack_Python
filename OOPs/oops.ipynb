```python
class Student():
    def __init__(self,name,marks,grade,percentage):
        self.name=name
        self.marks=marks
        self.grade=grade
        self.__percentage=percentage
    def get_percentage(self):
        return self.__percentage
    
    def show_data(self):
        print(f"Hi {self.name} your marks is {self.marks} and your grade is {self.grade} you got {self.__percentage}")
s1=Student("Madhav",650,"A",79)
s1.show_data()
s1.get_percentage()
```

```python
class Car:
    def __init__(self):
        self.__speed = 0

    def accelerate(self):
        self.__speed += 10
        print("Speed:", self.__speed)

    def brake(self):
        self.__speed -= 10
        print("Speed:", self.__speed)
```

```python
class Student:
    def __init__(self,name):
        self.name=name
        self.__marks=0
    def set_marks(self,marks):
        if marks>=0 and marks <=100:
            self.__marks=marks
        else:
            print("Invalid Marks")
    def get_marks(self):
        return self.__marks
s1=Student("Vishnu")
s1.set_marks(1)
s1.name
s1.get_marks()
```
